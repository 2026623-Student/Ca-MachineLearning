after checking just we found missing values in 5 variables, so next step is how we can handle it. 

```python
columns = ['ALB', 'ALP', 'ALT', 'CHOL', 'PROT']

for col in columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
df.isnull().sum()
```

Now we have 0 null values, next step is encoding our categorical variables in this case, Sex and Category

# Encoding

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df["Category"] = le.fit_transform(df["Category"])
df["Sex"] = le.fit_transform(df["Sex"])
```
